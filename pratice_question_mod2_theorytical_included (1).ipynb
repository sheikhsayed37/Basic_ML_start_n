**1. Countplot এবং Barplot এর মধ্যে পার্থক্য কী? কখন কোনটা use করবে?**



Answer :
  Countplot automatically কতবার data এসেছে সেটা count করে দেখায়।

  Mainly categorical data frequency দেখানোর জন্য ব্যবহার হয়।

  Example:  sns.countplot(data=df, x='gender')

  Barplot already থাকা numerical value এর average/mean বা অন্য statistical value দেখায়।

  এখানে usually:

  X = category
  
  Y = numerical value

  Example:  sns.barplot(data=df, x='department', y='score')



**2. Scatterplot দেখে কীভাবে বুঝবে দুটো variable এর মধ্যে সম্পর্ক আছে কিনা?**

Scatterplot ব্যবহার করা হয় দুইটি numerical variable এর মধ্যে relationship আছে কিনা দেখার জন্য।

Upward pattern   →  Positive relationship

Downward pattern →  Negative relationship

Random pattern   →  No relationship

Example: height vs weight, attendance vs score.

**3. Y Data Profiling কেন manually EDA করার চেয়ে সুবিধাজনক? কী কী information এক জায়গায় পাওয়া যায়?**

YData Profiling automatically পুরো dataset analysis করে দেয়, তাই manually EDA করার চেয়ে দ্রুত ও সহজ।

এক জায়গায় যেসব information পাওয়া যায়:
Missing values
Duplicate rows
Data types
Mean, median, max, min
Correlation between variables
Distribution of data
Outliers
Unique values

এতে খুব কম code লিখেই dataset সম্পর্কে overall idea পাওয়া যায়।_তোমার উত্তর এখানে লেখো_

**4. নিচের code টি একটি Countplot তৈরি করবে যেখানে passenger class অনুযায়ী count দেখাবে এবং Gender দিয়ে আলাদা করবে।**

*Hint: x = ticket class column, hue = gender column*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
print(df)
sns.countplot(data=df, x='pclass', hue='sex')

plt.title("Passenger Class by Gender")
plt.show()
```

**5. নিচের code টি City অনুযায়ী average Salary দেখাবে।**

*Hint: Cat-Num relationship দেখানোর plot, x = category, y = number*

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "City": ["Dhaka", "Dhaka", "CTG", "CTG", "Sylhet"],
    "Salary": [50000, 60000, 45000, 55000, 40000]
})
sns.barplot(data=df, x='City', y='Salary')

plt.title("Average Salary by City")
plt.show()
```

**6. নিচের code টি Age column এর smooth distribution curve দেখাবে।**

*Hint: Histogram এর smooth version যে plot দেখায়*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

sns.histplot(data=df, x='age',bins=50)

plt.title("Age Distribution")

plt.show()
```

**7. নিচের code টি Height এবং Weight এর মধ্যে সম্পর্ক দেখাবে।**

*Hint: দুটো numerical variable এর মধ্যে relationship দেখানোর plot*

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "Height": [150, 160, 170, 175, 180],
    "Weight": [50, 60, 70, 75, 80]
})

sns.scatterplot(data=df, x='Height', y='Weight')

plt.title("Height vs Weight")
plt.show()
```
