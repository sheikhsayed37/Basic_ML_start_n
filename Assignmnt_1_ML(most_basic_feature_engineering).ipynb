## **Question 10 (10 Marks)**

Handle missing values in the Titanic dataset by:

Imputing missing Age values using SimpleImputer with mean strategy

Imputing missing Embarked values using most frequent strategy

Imputing missing Cabin values using constant strategy with "Missing"

Finally, check whether any missing values remain.


```python
# x_train.isnull().sum()
age_imputer=SimpleImputer(missing_values=np.nan,strategy='mean')
age_imputer.fit(x_train[['Age']])

x_train['Age']=age_imputer.transform(x_train[['Age']])
x_test['Age']=age_imputer.transform(x_test[['Age']])

# test data teke noton mean ber kore nai

# train er mean ta use korse

# atai correct ML practice।
```

```python
print(x_train)
print(x_test)

```

```python
Embarked_imputer=SimpleImputer(missing_values=np.nan,strategy='most_frequent')
Embarked_imputer.fit(x_train[['Embarked']])

x_train['Embarked']=Embarked_imputer.transform(x_train[['Embarked']]).ravel()
x_test['Embarked']=Embarked_imputer.transform(x_test[['Embarked']]).ravel()
print(x_train)
print(x_test)

# test data teke noton frequent  ber kore nai

# train er frequent  ta use korse

```

```python
Cabin_imputer=SimpleImputer(missing_values=np.nan,strategy='constant',fill_value='Missing')
Cabin_imputer.fit(x_train[['Cabin']])
x_train['Cabin']=Cabin_imputer.transform(x_train[['Cabin']]).ravel()
x_test['Cabin']=Cabin_imputer.transform(x_test[['Cabin']]).ravel()
print(x_train)
print(x_test)

# test data teke noton kico use kore nai

# train er ta missing e convert kore seta  use kore dise test e


```

```python
print(x_train.isnull().sum())
print(x_test.isnull().sum())
```

# **Age column:**
1.The mean strategy was used because Age is a numerical feature. Replacing   missing values with the average age helps preserve the overall distribution of the data and is a common technique for handling missing numerical values.

# **Embarked column:**

The most_frequent strategy was applied because Embarked is a categorical feature containing port names. Filling missing values with the most common category ensures consistency and avoids introducing invalid values.

# **Cabin column:**
The constant strategy with the value "Missing" was used because a large number of cabin entries are absent. Instead of estimating them, a separate category called "Missing" was created to explicitly indicate unavailable cabin information.

After performing all imputations, the dataset was checked again to confirm that no missing values remained.
